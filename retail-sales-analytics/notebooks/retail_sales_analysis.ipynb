# Retail Sales & Customer Behavior Analysis
Synthetic dataset; descriptive analysis only.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
df = pd.read_csv(root / "outputs/cleaned_sales.csv", parse_dates=["order_date"])
df.head()
```

```python
df.info()
df[["revenue", "profit", "quantity"]].describe()
```

```python
monthly = df.groupby("month", as_index=False)[["revenue", "profit"]].sum()
sns.lineplot(data=monthly, x="month", y="revenue", marker="o")
plt.xticks(rotation=45); plt.title("Monthly revenue"); plt.tight_layout(); plt.show()
```

```python
by_category = df.groupby("category")[["revenue", "profit"]].sum().sort_values("revenue", ascending=False)
display(by_category)
by_category["revenue"].plot.bar(title="Revenue by category"); plt.tight_layout(); plt.show()
```

```python
customers = df.groupby("customer_id").agg(total_spend=("revenue", "sum"), orders=("order_id", "nunique"))
display(customers.sort_values("total_spend", ascending=False).head(10))
```
